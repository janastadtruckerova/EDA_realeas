# Exploratory Data Analysis - California Housing Dataset

## 1. Introduction
This notebook performs Exploratory Data Analysis (EDA) on the California Housing dataset. The goal is to explore the data, discover basic statistics, visualize the distribution of variables, and understand relationships between different factors affecting housing prices.


## 2. Import Libraries and Load Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Dataset Overview

In this section, we will take a first look at the dataset to understand its basic structure, contents, and characteristics. This initial exploration will help us identify potential challenges, including missing values, and inform our subsequent data preprocessing and analysis steps.

We will focus on the following aspects:

* **Loading the Dataset:** We will load the dataset into a pandas DataFrame, which is a fundamental data structure for data manipulation in Python.
* **Basic Information:** We will use the `.info()` method to get a concise summary of the DataFrame, including the number of rows and columns, data types of each column, and importantly, the number of non-null values. This will allow us to quickly identify columns with missing data.
* **Sample of the Data:** We will display the first few rows of the DataFrame using the `.head()` method to get a visual understanding of the data and the meaning of each column. We might also look for visual cues of missing data (e.g., `NaN` values).
* **Descriptive Statistics:** We will use the `.describe()` method to generate descriptive statistics of the numerical columns, such as count, mean, standard deviation, minimum, maximum, and quartiles. The 'count' statistic will also highlight the number of non-missing values in each numerical column.
* **Missing Value Analysis:** We will explicitly check for missing values using methods like `.isnull().sum()`. This will provide a clear count of missing values per column. We might also visualize the missing data patterns using libraries like `missingno` if needed for a more comprehensive understanding.
* **Unique Values:** For categorical columns, we might explore the number of unique values to understand the variety within each category.

By performing these initial steps, we aim to gain a foundational understanding of the dataset, including the extent and distribution of missing values, before diving into more detailed analysis and preprocessing strategies to handle them.



```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

# Load the dataset
california = fetch_california_housing()

# Convert the data to a pandas DataFrame
data = california.data
columns = california.feature_names
california_df = pd.DataFrame(data, columns=columns)

# Add the target variable (price) to the DataFrame
california_df['price'] = california.target

# Display the first few rows to inspect the data
california_df.head()
```

```python
# Basic information about the dataset
california_df.info()

# Summary statistics for numerical columns
california_df.describe()
```

```python
# Check for missing values
california_df.isnull().sum()

# Percentage of missing values
california_df.isnull().mean() * 100
```

## 3. Exploratory Data Analysis - Numerical Features

In this section, we will delve deeper into the numerical features of the California Housing dataset through visualization. Our goal is to understand the distribution of individual features and the relationships between them.

We will focus on the following visualizations:

* **Histograms of Numerical Columns:** Histograms are a powerful tool for visualizing the distribution of individual numerical features. They allow us to observe the frequency of different value ranges, identify potential skewness, and detect outliers. For each numerical column in the dataset, we will generate a histogram to understand its underlying distribution.

* **Correlation Matrix:** A correlation matrix is a table showing correlation coefficients between sets of variables. In our case, it will display the pairwise linear relationships between all numerical features. The correlation coefficient ranges from -1 to 1, where:
    * 1 indicates a perfect positive linear relationship.
    * -1 indicates a perfect negative linear relationship.
    * 0 indicates no linear relationship.

    Visualizing the correlation matrix using a heatmap can provide an intuitive understanding of which features are strongly correlated with each other. This can be useful for feature selection and understanding potential multicollinearity issues.

* **Scatter Plots:** Scatter plots are used to visualize the relationship between two numerical variables. Each point on the plot represents a single data instance, with its position determined by the values of the two chosen variables. By examining scatter plots of potentially correlated features (identified from the correlation matrix or based on domain knowledge), we can gain further insights into the nature and strength of their relationship. We might also focus on the relationship between the features and the target variable ('MedHouseVal') to understand which features have the most apparent linear influence on housing prices.

Through these visualizations, we aim to:

* Understand the distribution of individual numerical features.
* Identify potential linear relationships between different features.
* Gain insights into how numerical features might influence the target variable (median house value).
* Detect potential outliers or unusual patterns in the data.

The insights gained from this exploratory data analysis will be crucial for subsequent steps like feature engineering, model selection, and interpretation of the results.

```python
# Histograms for numerical columns
california_df.hist(figsize=(12, 10), bins=50)
plt.show()
```

```python
# Correlation matrix
corr_matrix = california_df.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.show()
```

```python
sns.scatterplot(x='AveRooms', y='price', data=california_df)
plt.show()
```
